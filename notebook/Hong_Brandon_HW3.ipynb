### (c) Feature Extraction

#### i. Research

Mean, Standard Deviation, Skewness, Kurtosis, Maximum, Minimum, Quantiles, Median

#### ii. Extraction

```python
all_column_names = []
for i in np.arange(1,7,1):
    all_column_names.extend([f'min_{i}', f'max_{i}', f'mean_{i}', f'median_{i}', f'std_{i}', f'1st_quart_{i}', f'3rd_quart_{i}'])
all_column_names
```

```python
extraction = []
for i in np.arange(0,88, 1):
    stats = instances[i].drop(columns = {'# Columns: time', 'instance_number'}).describe().T[['min', 'max', 'mean', '50%', 'std', '25%', '75%']].rename(columns = {'50%':'median', '25%': '1st_quart', '75%': '3rd_quart'}).to_numpy().flatten()
    extraction.append(stats)
time_series_stats = pd.DataFrame(extraction, columns= all_column_names)
time_series_stats.insert(0, 'Instance', np.arange(1,89,1))
time_series_stats
```

#### iii. Standard Deviation

```python
time_series_stats.drop(columns = {'Instance'}).std()
```

```python
length = time_series_stats.shape[0]
bootstrap_std = {}

for i in time_series_stats.columns[1:]:
    bootstrap_std[f'{i}_bootstrap_std'] = []
    for j in np.arange(0,1000, 1):
        sample_std = time_series_stats[f'{i}'].sample(n = length, replace = True).std()
        bootstrap_std[f'{i}_bootstrap_std'].append(sample_std)

bootstrap_std
```

```python
confidence_intervals = {}

for i in time_series_stats.columns[1:]:
    values = bootstrap_std[f'{i}_bootstrap_std']
    confidence_intervals[f'{i} Bootstrap 90% CI for STD'] = np.percentile(values, [5, 95])

confidence_intervals
```

## 2. ISLR 3.7.4

### (a) Linear Train

We would expect the training RSS for the cubic regression to be lower than the training RSS for the linear regression. Since the cubic model includes the linear model, the cubic model can become the linear model by setting the other variables to 0. Therefore, it will have the same or lower RSS. The cubic model may also try to account for the noise with its other variables, which would result in a lower training RSS.

### (b) Linear Test

We would expect the testing RSS for the linear regression to be smaller because the true relationship is linear.

### (c) Not Linear Train

The result is still the same. Although we may not know the true relationship anymore, the cubic model still contains the linear model. Therefore, we know it will perform as well as the linear model, but it may perform even better depending on the noise. The cubic model will be more flexible, so it will have a lower RSS. 

### (d) Not Linear Testing

Without knowing the true relationship, we do not have enough information to tell which testing RSS would be smaller. If the true relationship was linear, the cubic model could overfit on the training and perform poorly. Or if the true relationship was non-linear, the cubic model could capture the more complex relationship and perform better on the test data. Without the true relationship, we cannot tell.
